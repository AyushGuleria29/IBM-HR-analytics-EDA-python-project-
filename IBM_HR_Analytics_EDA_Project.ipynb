## <div style="text-align: left; background-color:#CFD9F1 ; font-family: Trebuchet MS; color: #0A3399; padding: 15px; line-height:1;border-radius:1px; margin-bottom: 0em; text-align: center; font-size: 25px;border-style: solid;border-color: dark green;"> 1. Visualizing the Employee Attrition Rate.</div> 

````python
plt.figure(figsize=(14,8))
plt.subplot(1,2,1)
attrition_rate = df["Attrition"].value_counts()
plt.pie(attrition_rate,labels=["Yes","No"],autopct="%.2f%%",textprops={"color":"black","size":15},colors=sns.color_palette("Set2"),explode=[0,0.1],startangle=90)
plt.title("Employee Attrition Rate",fontweight="black",color="red")
plt.tight_layout()
plt.show()
````

## Insights

- The attrition rate is at 16% of the organisation main reason for this attrition is low pay scale and overtime burden.
- The organisation needs to take necessary steps tp stop this attrtion rate going higher.



---

## <div style="text-align: left; background-color:#CFD9F1 ; font-family: Trebuchet MS; color: #0A3399; padding: 15px; line-height:1;border-radius:1px; margin-bottom: 0em; text-align: center; font-size: 25px;border-style: solid;border-color: dark green;"> 2. Analyzing Employee Attrition by Gender. </div> 

````python
plt.figure(figsize=(12,4))
att_gen = df["Gender"].value_counts()
plt.pie(att_gen.values,labels=att_gen.index,autopct="%.1f%%",startangle=90,textprops={"color":"black","size":15},wedgeprops={"width":0.5})
plt.title("Attrition by Gender",fontweight="black",size=15)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,4))
plt.subplot(1,2,2)
new_df = df[df["Attrition"]=="Yes"]
value_2 = new_df["Gender"].value_counts()
attrition_rate = np.floor((value_2/att_gen)*100).values
sns.barplot(x=value_2.index.tolist(),y= value_2.values,palette=("Set2"))
plt.title("Attrition Rate for Gender",fontweight="black",size=18)
for index,value in enumerate(value_2.values):
    plt.text(index,value, str(value)+" ("+str(attrition_rate[index])+"%)",ha="center",va="bottom",fontweight="black",size=10)

plt.tight_layout()
plt.show()
````

----

## Insights
- The number of male employees is 20 percent higher than no of females and also the attrition rate of males are higher than females in th organisation.


## <div style="text-align: left; background-color:#CFD9F1 ; font-family: Trebuchet MS; color: #0A3399; padding: 15px; line-height:1;border-radius:1px; margin-bottom: 0em; text-align: center; font-size: 25px;border-style: solid;border-color: dark green;"> 3. ## <div style="text-align: left; background-color:#CFD9F1 ; font-family: Trebuchet MS; color: #0A3399; padding: 15px; line-height:1;border-radius:1px; margin-bottom: 0em; text-align: center; font-size: 25px;border-style: solid;border-color: dark green;"> 5. Analyzing Employee Attrition by DailyRate. </div> 

* **Note:**
    1. **DailyRate** shows the Daily salary rate for employees.
    2. To **generate meaningful insights** we can cut **Daily Rates** into three groups for meaningful analysis.

---

````python
df["daily_rt"] = pd.cut(df["DailyRate"],bins=[100,500,1000,2000],labels=["Low","Medium","High"])
plt.figure(figsize=(14,6))
att_rate = df["daily_rt"].value_counts()
plt.pie(att_rate.values,labels=att_rate.index,autopct="%.1f%%",startangle=90,textprops={"color":"black","size":10},explode=[0,0.1,0.1])
plt.title("Employees Daily rate Group",fontweight="black",size=16,color="black")
plt.show()


plt.subplot(1,2,2)
new_df = df[df["Attrition"]=="Yes"]
value_2 = new_df["daily_rt"].value_counts()
attrition_rate = np.floor((value_2/att_rate)*100).values
sns.barplot(x=value_2.index.tolist(),y= value_2.values,palette=["#11264e","#6faea4","#FEE08B","#D4A1E7","#E7A1A1"])
plt.title("Attrition Rate by Daily rate",fontweight="black",pad=15,size=18)
for index,value in enumerate(value_2.values):
    plt.text(index,value, str(value)+" ("+str(attrition_rate[index])+"%)",ha="center",va="bottom",fontweight="black",size=10)

plt.tight_layout()
plt.show()
````

## Insights 
- The attrition rate of both medium and high are almost same as compared to daliy rate.
- 

## <div style="text-align: left; background-color:#CFD9F1 ; font-family: Trebuchet MS; color: #0A3399; padding: 15px; line-height:1;border-radius:1px; margin-bottom: 0em; text-align: center; font-size: 25px;border-style: solid;border-color: dark green;"> 6. Analyzing Employee Attrition by Distance From Home. </div> 

````python
df["dist_att"] = pd.cut(df["DistanceFromHome"],bins=[0,2,5,10,20],labels=["0-2km","3-5km","6,10km","10-20km"])

plt.figure(figsize=(13,6))
plt.subplot(1,2,1)
value_1 = df["dist_att"].value_counts()
sns.barplot(x=value_1.index.tolist(), y=value_1.values,palette="Set2")
plt.title("Employees Distance From Home",fontweight="black",size=15)
for index, value in enumerate(value_1.values):
    plt.text(index,value,value,ha="center",fontweight="black",size=15)
plt.show()    

plt.figure(figsize=(12,4))
plt.subplot(1,2,2)
new_df = df[df["Attrition"]=="Yes"]
value_2 = new_df["dist_att"].value_counts()
attrition_rate = np.floor((value_2/value_1)*100).values
sns.barplot(x=value_2.index.tolist(),y= value_2.values,palette="Set2")
plt.title("Attrition Rate for  DistanceFromHome",fontweight="black",pad=15,size=18)
for index,value in enumerate(value_2.values):
    plt.text(index,value, str(value)+" ("+str(attrition_rate[index])+"%)",ha="center",fontweight="black",size=10)

plt.tight_layout()
plt.show()
````

----

### insights
- The employees lives with very close distance to the organisation has higher attrition rate while the employees which lives  from 6- 10 km far from the organisation has lower no of attrition rate.
- The close distance employees leaves the organisation because they dont get the basic allowance like cab facilities fuel stifund etc.

# <div style="padding:20px;color:white;margin:0;font-size:35px;font-family:Verdana;text-align:center;display:fill;border-radius:5px;background-color:#254E58;overflow:hidden"><b>💡 Recommendations 💡</b></div>

---

#### Revaluate Payroll:
- Establish Entry-Level Minimums: Address the wide income disparity by re-evaluating the salary floor for Job Level 1 roles (currently starting as low as 1,000/month). Raising the baseline entry-level pay will directly curb early-tenure attrition.

-  Review employees nearing the 2-year mark in low-paying roles. Early, structured step-increases or milestone bonuses can mitigate the financial incentive to seek external opportunities.

### Career Path to counter role stagnation
- The "4-Year" Tenure Reset: With employees averaging roughly 4 years in their current roles and under the same managers, this threshold is a critical friction point. Implement mandatory career progression reviews or lateral rotation opportunities at the 2-3 year mark to prevent stagnation.
#### Rotational Programs
- We can introduce leadership rotation or cross-department mentorship programs. This exposes talent to new perspectives and internal opportunities       without requiring them to leave the company.
#### Overtime and travel
- Compensatory Time Off: Implement a formal "compensatory-time" policy specifically for the Research & Development department, ensuring employees who log heavy overtime are mandated to take restorative time off.
